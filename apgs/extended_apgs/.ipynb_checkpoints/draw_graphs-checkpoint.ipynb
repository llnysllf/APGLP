```python
import pandas as pd
import matplotlib.pyplot as plt
import os
import re


base_data = []
base_f_data = []

#filter outliers based on IQR
def filter_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]

for file in os.listdir():
    if file.endswith('.csv'):

        #get node number
        if file[4].isdigit():
            node_num = int(file[3:5])
        else:
            node_num = int(file[3])
       
        df = pd.read_csv(file)

        #filter outliers
        df_filtered = filter_outliers(df, 'Gurobi_runtime/microseconds')

        #average runtime from filtered data
        avg_runtime = df_filtered['Gurobi_runtime/microseconds'].mean()

        #store data separately
        if 'f' in file:
            base_f_data.append((node_num, avg_runtime))
        else:
            base_data.append((node_num, avg_runtime))

#sort by node number
base_data.sort()
base_f_data.sort()

#convert to DataFrames
df_base = pd.DataFrame(base_data, columns=['nodes', 'Gurobi_runtime/microseconds'])
df_base_f = pd.DataFrame(base_f_data, columns=['nodes', 'Gurobi_runtime/microseconds'])

plt.figure(figsize=(10, 5))

#base
plt.plot(df_base['nodes'], df_base['Gurobi_runtime/microseconds'], marker='o',color='black')
plt.xlabel('Node_num')
plt.ylabel('Gurobi_runtime/ µm')
plt.tight_layout()
plt.savefig('gurobi.png')
plt.show()
plt.close()

#base_fixed
plt.plot(df_base_f['nodes'], df_base_f['Gurobi_runtime/microseconds'], marker='o', color='black')
plt.xlabel('Node_num')
plt.ylabel('Gurobi_runtime/ µm')
plt.tight_layout()
plt.savefig('gurobi_fixed.png')
plt.show()
plt.close()
```
