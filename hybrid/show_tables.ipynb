```python
import os
import re
import pandas as pd

base_rows = []
fixed_rows = []

for file in os.listdir():
    if file.endswith('.csv'): 
        node_num = int(file[3])
        df = pd.read_csv(file)
        df.drop(df.columns[df.columns.str.contains('unnamed',case = False)],axis = 1, inplace = True)

        optimal_dict = {}
        qpu_dict = {}
        
        for col in df.columns:
            
            if not col[-1].isdigit():
                total = 3
            elif col[-2].isdigit():
                total = int(col[-2:])
            else:
                total = int(col[-1])
            
            #compute the percentage of 'Y'.
            if 'optimal' in col:
                y_count = (df[col] == 'Y').sum()
                optimal_dict[total] = round((y_count / len(df)) * 100, 2)

            elif 'qpu' in col:
                qpu_dict[total] = df[col].mean()

            
        total_tested = optimal_dict.keys()

        #for each time_limit, add a row to the appropriate table
        for t in sorted(total_tested):
            row = {
                'Nodes': node_num,
                'Time_limit': int(t)*10**6,
                'Average QPU Time': round(qpu_dict[t]),
                'Optimal %': optimal_dict[t]
            }
            if 'f' in file:
                fixed_rows.append(row)
            else:
                base_rows.append(row)

df_base = pd.DataFrame(base_rows).sort_values(by=['Nodes', 'Time_limit']).reset_index(drop=True)
df_fixed = pd.DataFrame(fixed_rows).sort_values(by=['Nodes', 'Time_limit']).reset_index(drop=True)

print("Unconstrained:")
display(df_base)  # Use display() in Jupyter Lab for a nicer view

print("\nFixed:")
display(df_fixed)
```
